```python
import pandas as pd

# 1. Descarregar o ficheiro de dados de Friends diretamente da internet
url = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-09-08/friends_info.csv'
df = pd.read_csv(url)

# 2. Visualizar as primeiras 5 linhas
df.head()
```

```python
# Calcular a nota média do IMDb por temporada
media_temporadas = df.groupby('season')['imdb_rating'].mean().reset_index()

# Renomear as colunas para ficar mais claro
media_temporadas.columns = ['Temporada', 'Nota_Media']

# Ordenar da temporada com maior nota para a menor
media_temporadas = media_temporadas.sort_values(by='Nota_Media', ascending=False)

print("--- NOTA MÉDIA POR TEMPORADA ---")
print(media_temporadas)
```

```python
# Ordenar episódios da maior nota para a menor
top_10_episodios = df.sort_values(by='imdb_rating', ascending=False)

# Mostrar os 10 episódios mais bem avaliados
top_10 = top_10_episodios[['season', 'episode', 'title', 'imdb_rating']].head(10)

print("--- TOP 10 EPISÓDIOS DE FRIENDS ---")
print(top_10)
```

```python
# 1. Instalar e importar a biblioteca para rodar SQL no Pandas
!pip install pandasql -q
from pandasql import sqldf

# Função auxiliar para rodar queries
pysqldf = lambda q: sqldf(q, globals())

# Query 1: Média de notas por temporada usando SQL
query_temporadas = """
SELECT
    season AS Temporada,
    ROUND(AVG(imdb_rating), 2) AS Nota_Media
FROM df
GROUP BY season
ORDER BY Nota_Media DESC;
"""

# Query 2: Top 5 episódios usando SQL
query_top_episodios = """
SELECT
    season AS Temporada,
    episode AS Episodio,
    title AS Titulo,
    imdb_rating AS Nota
FROM df
ORDER BY imdb_rating DESC
LIMIT 5;
"""

print("--- MÉDIA POR TEMPORADA (SQL) ---")
print(pysqldf(query_temporadas))

print("\n--- TOP 5 EPISÓDIOS (SQL) ---")
print(pysqldf(query_top_episodios))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar estilo visual do gráfico
plt.figure(figsize=(10, 5))
sns.barplot(data=media_temporadas, x='Temporada', y='Nota_Media', palette='plasma')

# Títulos e rótulos
plt.title('Nota Média do IMDb por Temporada de Friends ☕', fontsize=14, fontweight='bold')
plt.xlabel('Temporada', fontsize=12)
plt.ylabel('Nota Média', fontsize=12)
plt.ylim(8, 9) # Ajuste do eixo Y para destacar as variações

plt.show()
```

```python
# Lista dos 6 personagens principais
personagens = ['Rachel', 'Monica', 'Phoebe', 'Joey', 'Chandler', 'Ross']

# Função para identificar qual personagem está no título do episódio
def identificar_personagem(titulo):
    encontrados = [p for p in personagens if p.lower() in str(titulo).lower()]
    return encontrados[0] if encontrados else 'Outros / Vários'

# Criar uma nova coluna no nosso dataset
df['personagem_destaque'] = df['title'].apply(identificar_personagem)

# Calcular a nota média dos episódios onde cada personagem é citado no título
media_por_personagem = df[df['personagem_destaque'] != 'Outros / Vários'].groupby('personagem_destaque')['imdb_rating'].agg(['mean', 'count']).reset_index()
media_por_personagem.columns = ['Personagem', 'Nota_Media_Episodios', 'Qtd_Episodios_No_Titulo']

# Ordenar pela maior nota média
media_por_personagem = media_por_personagem.sort_values(by='Nota_Media_Episodios', ascending=False)

print("--- NOTA MÉDIA DOS EPISÓDIOS CENTRADOS EM CADA PERSONAGEM ---")
print(media_por_personagem)
```

```python
query_personagens = """
SELECT
    personagem_destaque AS Personagem,
    COUNT(*) AS Total_Episodios,
    ROUND(AVG(imdb_rating), 2) AS Nota_Media
FROM df
WHERE personagem_destaque != 'Outros / Vários'
GROUP BY personagem_destaque
ORDER BY Nota_Media DESC;
"""

print("--- ANÁLISE POR PERSONAGEM (SQL) ---")
print(pysqldf(query_personagens))
```

```python
import pandas as pd

# 1. Carregar o dataset completo de falas/diálogos de Friends
url_falas = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-09-08/friends.csv'
df_falas = pd.read_csv(url_falas)

# 2. Filtrar apenas pelos 6 personagens principais
personagens_principais = ['Rachel Green', 'Monica Geller', 'Phoebe Buffay', 'Joey Tribbiani', 'Chandler Bing', 'Ross Geller']
df_main = df_falas[df_falas['speaker'].isin(personagens_principais)]

# 3. Contar o total de falas por personagem
ranking_falas = df_main['speaker'].value_counts().reset_index()
ranking_falas.columns = ['Personagem', 'Total_de_Falas']

print("--- RANKING DE PERSONAGENS COM MAIS FALAS ---")
print(ranking_falas)
```

```python
query_falas = """
SELECT
    speaker AS Personagem,
    COUNT(*) AS Total_Falas,
    ROUND(COUNT(*) * 100.0 / (SELECT COUNT(*) FROM df_main), 2) AS Percentual_Do_Total
FROM df_main
GROUP BY speaker
ORDER BY Total_Falas DESC;
"""

print(pysqldf(query_falas))
```

```python
from itertools import combinations
from collections import Counter

# 1. Mapear nomes completos para nomes curtos
nomes_curtos = {
    'Rachel Green': 'Rachel',
    'Monica Geller': 'Monica',
    'Phoebe Buffay': 'Phoebe',
    'Joey Tribbiani': 'Joey',
    'Chandler Bing': 'Chandler',
    'Ross Geller': 'Ross'
}

# 2. Filtrar e aplicar nomes curtos
df_main = df_falas[df_falas['speaker'].isin(nomes_curtos.keys())].copy()
df_main['personagem'] = df_main['speaker'].map(nomes_curtos)

# 3. Agrupar os personagens presentes na mesma cena (usando a coluna 'scene')
cenas = df_main.groupby(['season', 'episode', 'scene'])['personagem'].unique()

# 4. Contar as combinações de duplas que aparecem na mesma cena
duplas_contador = Counter()

for lista_personagens in cenas:
    if len(lista_personagens) >= 2:
        for dupla in combinations(sorted(lista_personagens), 2):
            duplas_contador[dupla] += 1

# 5. Criar a Tabela Final
df_duplas = pd.DataFrame(duplas_contador.most_common(10), columns=['Dupla', 'Cenas_Juntos'])
df_duplas['Dupla'] = df_duplas['Dupla'].apply(lambda x: f"{x[0]} & {x[1]}")

print("--- TOP 10 DUPLAS COM MAIS CENAS JUNTAS ---")
print(df_duplas)
```
